```python
import requests as req
import json
import pandas as pd
from pprint import pprint
```

```python
info_endpoint = lambda n: f"https://pokeapi.co/api/v2/pokemon/{n}/"
rarity_endpoint = lambda n: f"https://pokeapi.co/api/v2/pokemon-species/{n}/"
```

#### CSV File Headers
Name, Generation, Type 1, Type 2, Ability 1, Ability 2, Hidden Ability, Legendary Status, HP, Atk, Def, Sp.Atk, Sp.Def, Spd, Total

```python
# Sample Request

fields_to_add = [1]     # Initializing the list with the National Pokedex ID
info_req = req.get(info_endpoint(1)).json()
rarity_req = req.get(rarity_endpoint(1)).json()


legend_or_myth = bool(rarity_req['is_legendary'] + rarity_req['is_mythical'])

fields_to_add += [info_req['name']]    # Name

fields_to_add += [rarity_req['generation']['url'][-2]]   # Generation
fields_to_add += [info_req['types'][i]['type']['name'] for i in range(len(info_req['types']))]   # Types

abilities = ['', '', '']
for ability_info in info_req['abilities']:
    abilities[ability_info['slot'] - 1] = ability_info['ability']['name']
fields_to_add += abilities     # Abilities

fields_to_add += [str(int(legend_or_myth))]      # Legendary Status

base_stats = {info_req['stats'][i]['stat']['name'] : info_req['stats'][i]['base_stat'] for i in range(len(info_req['stats']))}

total = 0
for k in base_stats.keys():
    total += base_stats[k]
base_stats['total'] = total

fields_to_add += [str(v) for k, v in base_stats.items()]   # Base Stats
```

```python
def generate_data(pkmn_id):
    fields_to_add = [pkmn_id]
    info_req = req.get(info_endpoint(pkmn_id)).json()
    rarity_req = req.get(rarity_endpoint(pkmn_id)).json()


    legend_or_myth = bool(rarity_req['is_legendary'] + rarity_req['is_mythical'])

    fields_to_add += [info_req['name']]    # Name

    fields_to_add += [rarity_req['generation']['url'][-2]]   # Generation
    
    types = ['', '']
    for type_info in info_req['types']:
        types[type_info['slot'] - 1] = type_info['type']['name']
    fields_to_add += types     # Types

    abilities = ['', '', '']
    for ability_info in info_req['abilities']:
        abilities[ability_info['slot'] - 1] = ability_info['ability']['name']
    fields_to_add += abilities     # Abilities

    fields_to_add += [str(int(legend_or_myth))]      # Legendary Status

    base_stats = {info_req['stats'][i]['stat']['name'] : info_req['stats'][i]['base_stat'] for i in range(len(info_req['stats']))}

    total = 0
    for k in base_stats.keys():
        total += base_stats[k]
    base_stats['total'] = total

    fields_to_add += [str(v) for k, v in base_stats.items()]   # Base Stats
    
    return fields_to_add
```

```python
import csv

def create_dataset():
    headers = "id,name,generation,type_1,type_2,ability_1,ability_2,hidden_ability,is_legendary,hp,atk,def,sp_atk,sp_def,speed,total".split(',')

    with open("pokemon.csv", "w", newline='') as csv_file:
        writer = csv.writer(csv_file)
        
        writer.writerow(headers)
        
        for i in range(1, 1025):
            data = generate_data(i)
            writer.writerow(data)
            print(f"Written data for National Dex No. {i} - {data[1].title()}")
```
